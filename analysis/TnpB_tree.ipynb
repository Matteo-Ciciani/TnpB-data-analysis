```python
import os
import pandas as pd
from Bio import Seq,SeqIO
```

```python
tnpb_table = pd.read_csv('../data/TnpB_variants.csv')
```

```python
tnpb_table = tnpb_table.dropna()
```

```python
tnpb_table
```

```python
candidate_seqs = [SeqIO.SeqRecord(Seq.Seq(seq), id=name) for name,seq in tnpb_table[
    ['Name', 'Seq']].values]
```

```python
reference_seqs = [rec for rec in SeqIO.parse('../data/Active_TnpBs.faa', 'fasta')]
```

```python
SeqIO.write(candidate_seqs+reference_seqs, 'to_align.faa', 'fasta')
```
